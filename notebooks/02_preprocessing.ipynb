```python
# ============================================
# 02 — PREPROCESSING & FEATURE ENGINEERING
# ============================================

import pandas as pd
import numpy as np

from src.data.load_data import load_dataset
from src.preprocessing.cleaning import clean_dataset
from src.preprocessing.feature_engineering import engineer_features
from src.preprocessing.encoding import encode_features
from src.preprocessing.scaling import scale_features

# --------------------------------------------
# 1. Load Raw Data
# --------------------------------------------

df = load_dataset("../data/raw/bank_transactions_data_2.csv")
df.head()
```

```python
# --------------------------------------------
# 2. Cleaning
# --------------------------------------------

df_clean = clean_dataset(df)
df_clean.head()
```

```python
# --------------------------------------------
# 3. Feature Engineering
# --------------------------------------------

df_feat = engineer_features(df_clean)
df_feat.head()
```

```python
# --------------------------------------------
# 4. Encoding Categorical Variables
# --------------------------------------------

df_encoded = encode_features(df_feat)
df_encoded.head()
```

```python
# --------------------------------------------
# 5. Scaling Numerical Features
# --------------------------------------------

df_scaled, scaler = scale_features(df_encoded)
df_scaled.head()
```

```python
# --------------------------------------------
# 6. Save Processed Data
# --------------------------------------------

df_clean.to_csv("../data/processed/cleaned.csv", index=False)
df_encoded.to_csv("../data/processed/encoded.csv", index=False)
df_scaled.to_csv("../data/processed/scaled.csv", index=False)

print("Phase 2 preprocessing completed successfully.") 
```
